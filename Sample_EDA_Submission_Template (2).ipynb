## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
## 1. Drop 'comments' column (86.9% missing, free-text, not useful for EDA)
df.drop(columns=['comments'], inplace=True, errors='ignore')

## 2. Drop 'Timestamp' (not analytically useful)
df.drop(columns=['Timestamp'], inplace=True, errors='ignore')

## 3. Clean Age column - remove outliers
df = df[(df['Age'] >= 18) & (df['Age'] <= 75)]
print(f"Shape after Age cleaning: {df.shape}")

## 4. Standardize Gender column
def clean_gender(g):
    g = str(g).strip().lower()
    if g in ['male', 'm', 'man', 'cis male', 'male (cis)', 'maile', 'mal', 'male-ish', 'make', 'msle', 'malr', 'cis man']:
        return 'Male'
    elif g in ['female', 'f', 'woman', 'cis female', 'femake', 'femail', 'cis-female/femme', 'female (cis)', 'female ']:
        return 'Female'
    else:
        return 'Other'

df['Gender'] = df['Gender'].apply(clean_gender)
print("Gender value counts:", df['Gender'].value_counts().to_dict())

## 5. Fill missing values
df['state'].fillna('Unknown', inplace=True)
df['self_employed'].fillna(df['self_employed'].mode()[0], inplace=True)
df['work_interfere'].fillna("Don't Know", inplace=True)

## 6. Verify no missing values remain
print(f"\nRemaining nulls: {df.isnull().sum().sum()}")
print(f"Final shape: {df.shape}")
````

### What all manipulations have you done and insights you found?

1. **Dropped `comments`** (86.9% missing, free-text — not suitable for structured EDA).
2. **Dropped `Timestamp`** (not analytically relevant to mental health patterns).
3. **Cleaned `Age`**: Removed records with age < 18 or > 75 (outliers like -1726 and 1e+11). Lost ~5 records.
4. **Standardized `Gender`**: Mapped 49 inconsistent entries into 3 clean categories: *Male*, *Female*, *Other*.
5. **Imputed missing values**:
   - `state`: Filled with `'Unknown'` since it's only applicable to US respondents.
   - `self_employed`: Filled with mode (majority were not self-employed).
   - `work_interfere`: Filled with `"Don't Know"` — absence of response likely means no mental health condition.

## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1 visualization code
# Treatment Rate Distribution (Pie Chart)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

treatment_counts = df['treatment'].value_counts()
axes[0].pie(treatment_counts, labels=treatment_counts.index, autopct='%1.1f%%',
            colors=['#66b3ff', '#ff9999'], startangle=90, explode=(0.05, 0))
axes[0].set_title('Overall Treatment Rate', fontsize=14)

# Treatment by Gender
treatment_gender = df.groupby('Gender')['treatment'].value_counts(normalize=True).unstack() * 100
treatment_gender['Yes'].plot(kind='bar', ax=axes[1], color=['#5bc0eb', '#fde74c', '#9bc53d'], edgecolor='black')
axes[1].set_title('Treatment Rate by Gender (%)', fontsize=14)
axes[1].set_xlabel('Gender')
axes[1].set_ylabel('% Seeking Treatment')
axes[1].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A pie chart clearly shows the overall proportion of treatment seekers. A grouped bar chart shows gender-wise comparison effectively, making it easy to spot disparities.

##### 2. What is/are the insight(s) found from the chart?

- About **50.7%** of respondents have sought mental health treatment — nearly half the surveyed tech workforce.
- **Female respondents** show a significantly higher treatment rate (75%) compared to **Males (~44%)** and **Other genders (~67%)**.
- This may reflect that women are more likely to recognize and seek help, or face higher stressors.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. The high overall treatment rate indicates a significant need for mental health support. Companies that proactively address this can reduce absenteeism and improve productivity.
The gender disparity may signal that male employees are **under-seeking** treatment — targeted awareness campaigns for male employees could bridge this gap.

#### Chart - 2

````python
# Chart - 2 visualization code
# Age Distribution by Treatment
plt.figure(figsize=(10, 5))
for treat in ['Yes', 'No']:
    subset = df[df['treatment'] == treat]['Age']
    subset.plot(kind='kde', label=treat, linewidth=2)
plt.title('Age Distribution by Treatment Status', fontsize=14)
plt.xlabel('Age')
plt.ylabel('Density')
plt.legend(title='Sought Treatment')
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

KDE (Kernel Density Estimate) plots are ideal for comparing continuous variable distributions across two groups without binning bias.

##### 2. What is/are the insight(s) found from the chart?

- Both groups show a similar age distribution peaking around **28–35 years**.
- Employees who sought treatment tend to be slightly **older on average**, suggesting mental health concerns accumulate with career tenure.
- The distributions largely overlap, indicating age alone is not a strong predictor of treatment-seeking.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Age-targeted interventions (e.g., onboarding mental health resources for new hires, burnout programs for mid-career employees) can be designed based on these distribution patterns.

#### Chart - 3

````python
# Chart - 3 visualization code
# Family History vs Treatment
cross = pd.crosstab(df['family_history'], df['treatment'])
cross_pct = cross.div(cross.sum(axis=1), axis=0) * 100

cross_pct.plot(kind='bar', figsize=(8, 5), colormap='Set2', edgecolor='black')
plt.title('Treatment Rate by Family History', fontsize=14)
plt.xlabel('Family History of Mental Illness')
plt.ylabel('Percentage (%)')
plt.xticks(rotation=0)
plt.legend(title='Treatment')
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A stacked/grouped bar chart of a crosstab is the clearest way to compare categorical ratios across two binary variables.

##### 2. What is/are the insight(s) found from the chart?

- **76%** of those with a family history of mental illness sought treatment vs. only **34%** without family history.
- Family history is one of the **strongest predictors** of treatment-seeking behavior in this dataset.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. Employees with family history are a high-risk group. Companies can design targeted support (counseling, EAP programs) for this segment through confidential intake surveys. This proactive approach can prevent crises before they occur.
